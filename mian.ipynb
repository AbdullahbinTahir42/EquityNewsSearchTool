```python
from langchain_community.document_loaders import UnstructuredURLLoader

loader = UnstructuredURLLoader(
    urls = [
        "https://info.deeplearning.ai/china-questions-nvidia-when-models-memorize-mixture-of-video-experts-openai-oracle-join-forces-1?ecid=ACsprvtqS8eKxR3R_Jkt3PgUKrlhfrffQSw60GMXMrPg2LST_Ap_N68ecxLZsRV3zThcpuo_A-95&utm_campaign=The%20Batch&utm_medium=email&_hsenc=p2ANqtz-95qbLcNQPMCnyRWs-gFrqLTibGWEUCci7JS2ho-NgjIbYVwcMXStR6p28ZayBTCSI77bSxyoqas7NUFKRz-YKKtCdhqRy68u2pAw8TWZiDDy1fgnY&_hsmi=376823600&utm_content=376818908&utm_source=hs_email",
        "https://www.geo.tv/latest/619885-indias-top-court-allows-release-of-picked-up-stray-dogs-in-delhi"
    ]
)
documents = loader.load()
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)

texts = text_splitter.split_documents(documents)
```

```python
len(texts)
text_contents = [doc.page_content for doc in texts]
```

```python
from sentence_transformers import SentenceTransformer

encoder = SentenceTransformer('all-mpnet-base-v2')


embeddings = encoder.encode(text_contents)
```

```python
from faiss import IndexFlatL2

index = IndexFlatL2(embeddings.shape[1])
index.add(embeddings)
```

```python
search_query = "Who hosted the Buildathon event mentioned in the text?"
```

```python
vec = encoder.encode(search_query)
vec.shape
```

```python
import numpy as np
s_vec = np.array(vec).reshape(1, -1)
s_vec.shape
```

```python
distance, I = index.search(s_vec, k=3)
I
```
